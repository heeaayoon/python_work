```python
# 직접 my_sort() 함수 짜기
l = [3,2,4,1]

def my_sort(l):
    for i in range(len(l)):
        for j in range(len(l)):
            if (l[i]<l[j]):
                l[i], l[j] = l[j], l[i]
    return l

my_sort(l)
```

```python
#list comprehension
data = [1,2,3]
new_data = []
for i in data:
    new_data.append(i*i)
print(new_data)

new = [i*i for i in data]
print(new)
```

```python
data = [1,2,3]
new_data = []
for i in data:
    if i%2 !=0:
        new_data.append(i*i)
print(new_data)

new = [i*i for i in data if i%2 !=0]
print(new)
```

```python
#list comprehension

def extract_vowel(s):
    result = []
    for i in s:
        if i in 'aeiou':
            result.append(i)
    return result

print(extract_vowel("hello"))

s= 'hello world'
def extract_vowel2(s):
    return [v for v in s if v in 'aeiou']

extract_vowel2(s)
```
