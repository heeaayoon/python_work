## 실습 문제 1: 일차 함수

함수 f(x) = 2x + 1 을 구현하고 그래프를 그려보세요.

- **요구사항:**

    0부터 5까지의 x 값에 대해 y 값을 계산하는 f(x) 함수를 순수 Python으로 작성하세요. (힌트: range 와 list 사용)



```python
import matplotlib.pyplot as plt

def f(x):
    # y= []
    # for i in x:
    #     y.append(2*i+1)
    # return y
    return [2*i+1 for i in x]


x = range(-5,6)
print(list(x))
y = f(x)
print(y)

plt.plot(x,y)
plt.grid()
```

## 실습 문제 2: 이차 함수 (다항 함수)

함수 f(x) = 2x^2 - 3x + 4 를 구현하고 그래프를 그려보세요.

- **요구사항:**

    0부터 5까지의 x 값에 대해 y 값을 계산하는 f(x) 함수를 순수 Python으로 작성하세요.

```python
import matplotlib.pyplot as plt
import numpy as np

def f(x):
    return (2*x)**2-(3*x)+4

x = range(1,6)
print(list(x))
y = f(x)
print(y)

plt.plot(x,y)
plt.grid()
```

```python
import matplotlib.pyplot as plt

def f(x):
    y= []
    for i in x:
        y.append((2*i)**2-(3*i)+4)
    return y

x = range(1,6)
print(list(x))
y = f(x)
print(y)

plt.plot(x,y)
plt.grid()
```

```python
# iterator 를 반환하는 함수들 
# product(*iterables) : 여러 개의 iterable한 객체에서 각각 하나씩 요소를 뽑아 만들 수 있는 모든 순서쌍(튜플)
# permutations(iterable, r) : 순열,  하나의 iterable에서 주어진 길이(r)만큼의 원소를 뽑아 나열 -> 순서가 다르면 다른 경우임
# combinations(iterable, r) : 조합,  하나의 iterable에서 주어진 길이(r)만큼의 원소를 뽑아 나열 -> 순서 고려 X
```

```python
from itertools import combinations

candidates = ["Alice", "Bob", "Charlie", "David"]
k=2

def get_teams(candidates: list, k: int) -> list:
    return list(combinations(candidates,k))
    
result = get_teams(candidates, k)    
print(result)
```

```python
from itertools import combinations
def get_teams(candidates: list, k: int) -> list:
    for c in candidates:
        result = list(combinations(c,k))
    return result


input_data = [
    (["Alice", "Bob", "Charlie", "David"], 2),
    (["A", "B", "C"], 3),
    (["X", "Y"], 1),
    (["Z"], 2)
]

output_data = [
    [('Alice', 'Bob'), ('Alice', 'Charlie'), ('Alice', 'David'), ('Bob', 'Charlie'), ('Bob', 'David'), ('Charlie', 'David')],
    [('A', 'B', 'C')],
    [('X',), ('Y',)],
    []
]

for (candidates, k), expected in zip(input_data, output_data):
    result = get_teams(candidates, k)
    #assert sorted(result) == sorted(expected), f"Failed: candidates={candidates}, k={k}, expected={expected}, got={result}"
    print(result)
print("Problem 1: All tests passed!")
```
