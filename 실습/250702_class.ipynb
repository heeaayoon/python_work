### class

```python
class Adder:
    def __init__(self,a,b):
        self.a = a
        self.b = b
    def sum(self):
        return self.a +self.b
    def __call__(self, *args, **kwds):
        return self.a +self.b
    def __str__(self):
        return str(self.a +self.b)
    
add = Adder(2,3)
add.sum()
```

```python
add()
```

```python
add = Adder(2,3)
print(add)
```

```python
class MySum:
    def __init__(self, a :list):
        self.data=a
    def __call__(self, *args, **kwds):
        return sum(self.data)
    def __getitem__(self, idx : int):
        return self.data[idx]
```

```python
s = MySum([1,2,3,4]) # __getitem__ 함수 호출됨 
s[0] +s[1]
```

```python
s = MySum([1,2,3,4]) #__call__ 함수 호출됨
s()
```

```python
class Range:
    def __init__(self, a :int):
        self.a = a
        self.current = 0
        
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.current < self.a:
            next_val =self.current+1
            self.current+=1
            return next_val
        else:
            raise StopIteration()



for i in Range(5):
    print(i)
```

```python
r = Range(5)
r_iter = iter(r)
print(next(r_iter))
print(next(r_iter))
print(next(r_iter))
print(next(r_iter))
print(next(r_iter))
print(next(r_iter))
```
