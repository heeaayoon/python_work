class inheritant

```python
class Number :
    def __init__(self, a):
        self.a = a

    def __add__(self, other):
        return Number(self.a + other.a)
    
    def __str__(self):
        return str(self.a)
    
    def __repr__(self):
        return self.__str__()
```

```python
n1 = Number(5)
n2 = Number(10)

n1+n2 #15
```

```python
class EvenNumber(Number):
    def __init__(self, a):
        super().__init__(a)
        if self.a%2!=0:
            self.a = self.a+self.a
```

```python
n1 = EvenNumber(5) #5+5
n2 = EvenNumber(10)

n1+n2 #20
```

```python
from abc import ABC, abstractmethod

class player(ABC):
    def __init__(self):
        super().__init__()
        self.name = str()

    @abstractmethod
    def run(self):
        pass
```

```python
class sonny(player):
    def __init__(self):
        super().__init__()
        self.name = "sonny"
```

```python
s = sonny()
s.run() # run 함수를 재정의 안했기 때문에 실행불가 
```
