```python
import matplotlib.pyplot as plt
from pathlib import Path
import csv
from datetime import datetime

path = Path('weather_data/sitka_weather_2021_simple.csv')
lines = path.read_text().splitlines()

reader = csv.reader(lines)
header = next(reader)
print(header)

#최고기온, 날짜, 최저기온 추출
dates, highs, lows = [], [], []
for row in reader:
    date = datetime.strptime(row[2],'%Y-%m-%d')
    high = int(row[4])
    low = int(row[5])
    highs.append(high)
    dates.append(date)
    lows.append(low)

fig,ax = plt.subplots()
ax.plot(dates,highs)
ax.plot(dates,lows)

plt.show()
```

```python
import matplotlib.pyplot as plt
from pathlib import Path
import csv
from datetime import datetime

path = Path('weather_data/death_valley_2021_simple.csv')
lines = path.read_text().splitlines()

reader = csv.reader(lines)
header = next(reader)
print(header)

#최고기온, 날짜, 최저기온 추출 -> 누락된 데이터 처리
dates, highs, lows = [], [], []
for row in reader:
    date = datetime.strptime(row[2],'%Y-%m-%d')
    try:
        high = int(row[3])
        low = int(row[4])
    except ValueError:
        print(f"{date}의 데이터 없음")
    else:
        highs.append(high)
        dates.append(date)
        lows.append(low)

fig,ax = plt.subplots()
ax.plot(dates,highs)
ax.plot(dates,lows)

plt.show()
```
