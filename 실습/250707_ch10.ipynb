```python
#10.1 open(파일이름, 파일 읽기 r / 쓰기 w) -> 직접열고 닫기
f = open('Numbers.txt','r', encoding='utf-8')
#f.read(5) #바이트 단위로 읽기
f.readline().rstrip() #줄 단위로 읽기 #rstrip() : 오른쪽 끝의 공백 문자를 제거합니다.
f.close()
```

```python
# with 블록을 넘어가면 자동으로 파일 닫기
with open('Numbers.txt','r', encoding='utf-8') as f:
    print(f.readline(),' ')
```

```python
# 학생별 평균
import csv
with open('grade.csv','r') as f:
    file = csv.reader(f)
    students = {}
    subjects = {}
    headers = next(file)
    for line in file:
        name = line[0]
        java = int(line[1])
        python = int(line[2])
        students[name] = sum([java,python])/2
        subjects['java'] = subjects.get('java', 0) + java
        subjects['python'] = subjects.get('python', 0) + python
students 
```

```python
# 과목별 평균 
for k,v in subjects.items():
    print(k,v / len(subjects))
```

```python
lst = [1,2,3]
contents = json.dumps(lst) # json.dumps(): Python 객체 → JSON 형식의 문자열(String)로 변환
```

```python
requset = json.loads(contents) # json.loads(): JSON 형식의 문자열(String) → Python 객체로 변환
type(requset), requset
```
