```python
def max_elements(a,b,c):
    return max(a,b,c)

max_elements(1,2,3)
```

```python
data = [1,2,3]
data = (1,2,3)
max_elements(*data)
```

```python
params = {'a':1, 'b':2, 'c':3 }

max_elements(**params)

# **params : 딕셔너리 언패킹 -> 값(value)을 해당 파라미터의 인자(argument)로 매핑하여 전달
# *params : 이터러블 언패킹 -> 이터러블의 각 요소를 순서대로 함수의 위치 인자(positional argument)로 전달
#                        -> 딕셔너리에 *를 사용하면, 딕셔너리의 키(key)들만 순서대로 언패킹/ 값(value)은 사용되지 않습니다.
```

```python
def max_elements2(a,b, *args):
    return max(a,b, *args)

max_elements2(1,2,3,4,5)
```

```python
def max_elements3(a, b, **kargs):
    return max(a,b, **kargs)


params = {'d':5, 'e':4, 'f':3 }
max_elements3(1,2,**params)
```

```python
def swap(a,b):
    a,b = b,a

a,b = 1,2 # 객체 아님
swap(a,b)
a,b
```

```python
def swap2(a,b):
    a[0],b[0] = b[0],a[0]

a,b = [1],[2] # 객체로 보기
swap2(a,b)
a,b
```
