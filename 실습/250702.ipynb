```python
s = 'hello, world!'

def count_alpha(str) -> dict:
    result = {}
    for s in str:
        #result[s] = result.get(s) +1
        result[s] = result.get(s,0) +1 #s가 없으면 0을 치환
    return result

count_alpha(s)
```

```python
from string import ascii_letters

s = 'hello, world!'

def count_alpha(str) -> dict:
    result = {}
    for s in str:
        if s in ascii_letters: #필터링
            result[s] = result.get(s,0) +1 #s가 없으면 0을 치환
    return dict(sorted(result.items(), key= lambda x:x[0])) # key 값으로 오름차순 정렬

count_alpha(s)
```
