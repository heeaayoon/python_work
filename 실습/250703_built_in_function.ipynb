```python
class Range:
    def __init__(self,n):
        self.n = n
        self.current = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < self.n:
           return_value = self.current
           self.current += 1
           return return_value
        raise StopIteration()
    
    def __len__(self):
        return self.n
    
    def __str__(self):
        return str('Range class')
    
    def __repr__(self):
        return self.__str__()
```
